# 6. Creación del modelo

Red densa (*feed-forward*) para regresión:

- **Capas ocultas:** `Dense` con activación **ReLU**. En el ejemplo del profesor se usa un número de neuronas similar al número de variables de entrada; aquí se prueban varias configuraciones.
- **Dropout (opcional):** apaga aleatoriamente un porcentaje de neuronas en cada paso de entrenamiento para reducir el sobreajuste.
- **Capa de salida:** **1 neurona sin activación** (lineal), porque se predice una sola variable continua: `quality`.
- **Pérdida:** `mse` (error cuadrático medio). **Optimizador:** `adam`. **Métrica adicional:** `mae`.

Se define una función para construir cada configuración y compararlas bajo las mismas condiciones. Se prueban **3 configuraciones** (sin Dropout, con Dropout y una red más profunda).

```python
def build_regression_model(hidden_units, dropout=0.0, learning_rate=0.001, name="modelo"):
    '''Construye una ANN de regresión.

    hidden_units : lista con el número de neuronas de cada capa oculta
    dropout      : tasa de Dropout después de cada capa oculta (0 = sin Dropout)
    '''
    model = keras.Sequential(name=name)
    model.add(layers.Input(shape=(X_train.shape[1],)))     # 12 variables de entrada
    for units in hidden_units:
        model.add(layers.Dense(units, activation="relu"))
        if dropout > 0:
            model.add(layers.Dropout(dropout))
    model.add(layers.Dense(1))   # 1 neurona de salida: la calidad (0-10)

    model.compile(optimizer=keras.optimizers.Adam(learning_rate=learning_rate),
                  loss="mse", metrics=["mae"])
    return model


# Configuraciones a probar
CONFIGS = {
    "A_sin_dropout":   {"hidden_units": [64, 32],      "dropout": 0.0},
    "B_dropout_0.2":   {"hidden_units": [64, 32],      "dropout": 0.2},
    "C_profunda_drop": {"hidden_units": [128, 64, 32], "dropout": 0.2},
}

build_regression_model(**CONFIGS["B_dropout_0.2"], name="ejemplo").summary()
```

# 7. Entrenamiento del modelo

- `validation_data`: en cada época se calcula la pérdida sobre el conjunto de validación (`val_loss`).
- `batch_size`: cantidad de datos que toma la red en cada cálculo (potencia de 2).
- `epochs`: máximo de iteraciones sobre todos los datos para ajustar los pesos (w) y sesgos (b).
- **`EarlyStopping`**: detiene el entrenamiento cuando `val_loss` deja de mejorar durante `patience` épocas y **restaura los pesos de la mejor época**. Por eso se puede fijar un número alto de épocas sin que la red se sobreajuste.

```python
EPOCHS = 300
BATCH_SIZE = 32
PATIENCE = 20

models, histories = {}, {}

for cfg_name, cfg in CONFIGS.items():
    keras.utils.set_random_seed(SEED)           # misma semilla para comparar de forma justa
    model = build_regression_model(**cfg, name=cfg_name.replace(".", "_"))

    early_stop = EarlyStopping(monitor="val_loss", patience=PATIENCE,
                               restore_best_weights=True, verbose=0)

    hist = model.fit(X_tr, y_tr,
                     validation_data=(X_val, y_val),
                     batch_size=BATCH_SIZE, epochs=EPOCHS,
                     callbacks=[early_stop], verbose=0)

    models[cfg_name], histories[cfg_name] = model, hist
    best_epoch = int(np.argmin(hist.history["val_loss"])) + 1
    print(f"{cfg_name:16s} | épocas ejecutadas: {len(hist.history['loss']):3d} "
          f"| mejor época: {best_epoch:3d} | mejor val_loss: {min(hist.history['val_loss']):.4f}")
```

```python
# Pérdidas del modelo (loss) y de validación (val_loss) por época, de la configuración B
losses = pd.DataFrame(histories["B_dropout_0.2"].history)
losses.tail()
```

### 7.1 Curvas de entrenamiento

```python
fig, axes = plt.subplots(1, len(CONFIGS), figsize=(5.2 * len(CONFIGS), 4), sharey=True)
for ax, (cfg_name, hist) in zip(axes, histories.items()):
    h = pd.DataFrame(hist.history)
    ax.plot(h["loss"], label="Entrenamiento")
    ax.plot(h["val_loss"], label="Validación")
    best = int(h["val_loss"].idxmin())
    ax.axvline(best, color="gray", linestyle="--", alpha=0.6, label=f"Mejor época ({best + 1})")
    ax.set_title(cfg_name); ax.set_xlabel("Época")
    ax.set_ylim(0, 2)   # zoom: la 1ª época (pérdida ~12) aplasta la escala
    ax.legend()
axes[0].set_ylabel("Pérdida (MSE)")
plt.suptitle("Modelo 1 (Regresión): curvas de pérdida", y=1.02)
plt.tight_layout()
plt.savefig(ProjectConfig.FIGURES_DIR / "modelo1_curvas_perdida.png", dpi=150, bbox_inches="tight")
plt.show()
# Idealmente las dos curvas bajan juntas. Si la de validación sube mientras la de
# entrenamiento sigue bajando, hay sobreajuste (EarlyStopping lo corta a tiempo).
```

# 8. Evaluación del modelo

Métricas de regresión (https://scikit-learn.org/stable/modules/model_evaluation.html#regression-metrics):

- **MAE** (error absoluto medio): en promedio, cuántos puntos de calidad se equivoca el modelo.
- **RMSE** (raíz del error cuadrático medio): como el MAE pero penaliza más los errores grandes.
- **R²** y **varianza explicada**: cuanto más cerca de 1, mejor (0 equivale a predecir siempre el promedio).

Como referencia se incluye un **modelo base** que siempre predice el promedio de calidad del entrenamiento: la red debe superarlo para tener sentido.

## 8.1 Predicción sobre el conjunto de Test

```python
def evaluate_regression(y_true, y_pred):
    y_pred = np.asarray(y_pred).ravel()
    return {
        "MAE":  mean_absolute_error(y_true, y_pred),
        "RMSE": float(np.sqrt(mean_squared_error(y_true, y_pred))),
        "R2":   r2_score(y_true, y_pred),
        "Var_explicada": explained_variance_score(y_true, y_pred),
    }

rows = []
# Línea base: predecir siempre la media
base_pred = np.full(len(y_test), y_train.mean())
rows.append({"Configuración": "Base (promedio)", **evaluate_regression(y_test, base_pred),
             "Épocas": 0, "Parámetros": 0})

predictions = {}
for cfg_name, model in models.items():
    predictions[cfg_name] = model.predict(X_test, verbose=0).ravel()
    val_mae = mean_absolute_error(y_val, model.predict(X_val, verbose=0).ravel())
    rows.append({
        "Configuración": cfg_name,
        **evaluate_regression(y_test, predictions[cfg_name]),
        "Épocas": len(histories[cfg_name].history["loss"]),
        "Parámetros": model.count_params(),
        "MAE_val": val_mae,
    })

results = pd.DataFrame(rows).set_index("Configuración")
results.round(4)
```

```python
# ¿En qué porcentaje nos estamos equivocando respecto al valor medio / mediana de la calidad?
for cfg_name in CONFIGS:
    mae = mean_absolute_error(y_test, predictions[cfg_name])
    print(f"{cfg_name:16s} | MAE / media: {mae / df['quality'].mean():.1%} | MAE / mediana: {mae / df['quality'].median():.1%}")
```

```python
# La mejor configuración se elige con la VALIDACIÓN (no con el test, para no "mirar" el examen final)
best_name = results.drop(index="Base (promedio)")["MAE_val"].idxmin()
best_pred = predictions[best_name]
print("Mejor configuración según validación:", best_name)
```

```python
# Visualizar nuestra predicción: valor real vs. predicción
errors = y_test.values - best_pred   # error = real - predicción

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

axes[0].scatter(y_test, best_pred, alpha=0.35, s=14)
axes[0].plot([3, 9], [3, 9], "r--", label="Predicción perfecta")
axes[0].set_xlabel("Valor real"); axes[0].set_ylabel("Predicción")
axes[0].set_title(f"Real vs. predicho ({best_name})"); axes[0].legend()

sns.histplot(errors, bins=30, kde=True, ax=axes[1], color="#55A868")
axes[1].axvline(0, color="red", linestyle="--")
axes[1].set_xlabel("Error (real - predicho)"); axes[1].set_title("Distribución del error")

plt.tight_layout()
plt.savefig(ProjectConfig.FIGURES_DIR / "modelo1_real_vs_predicho.png", dpi=150, bbox_inches="tight")
plt.show()
# Idealmente el error debe estar concentrado en 0. Los puntos se agrupan en columnas
# verticales porque la calidad real es un número entero y la predicción es continua.

print(f"Media del error: {errors.mean():+.3f} | Desv. estándar: {errors.std():.3f}")
print(f"Predicciones dentro de ±0.5 puntos: {np.mean(np.abs(errors) <= 0.5):.1%}")
print(f"Predicciones dentro de ±1.0 punto : {np.mean(np.abs(errors) <= 1.0):.1%}")
```

```python
# ¿Dónde falla más el modelo? Error promedio según la calidad real
err_by_q = (pd.DataFrame({"calidad_real": y_test.values, "error_abs": np.abs(errors)})
            .groupby("calidad_real")["error_abs"].agg(["mean", "count"]).round(3))
err_by_q.columns = ["MAE", "n_vinos"]
err_by_q
```
